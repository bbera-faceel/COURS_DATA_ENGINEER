# Cours 1 : Fondations de la manipulation de données avec Pandas
## Chapitre 1. Introduction aux Series et DataFrames

En Data Engineering, on manipule rarement la donnée avec de simples boucles Python et des dictionnaires. Pandas apporte des structures optimisées en C (vectorisation, typage strict) :
- **`pd.Series`** : Tableau 1D étiqueté, homogène en type.
- **`pd.DataFrame`** : Tableau 2D étiqueté (collection de Series partageant le même index).

```python
import pandas as pd
import numpy as np
```

### 1.1 Découverte des Series
Une Series possède un index (labels) et un type de données (`dtype`).

```python
# Création d'une Series avec index personnalisé
temperatures = pd.Series([18.5, 21.0, 19.2, 24.5], index=["Lille", "Paris", "Lyon", "Marseille"], name="temp_celsius")
print(temperatures)
print("\nType des données :", temperatures.dtype)
print("Accès par label (Lille) :", temperatures["Lille"])
print("Calcul vectorisé (+2°C) :\n", temperatures + 2)
```

### 1.2 Découverte des DataFrames
Un DataFrame peut être vu comme une table SQL ou une feuille Excel. Chaque colonne est une `Series`.

```python
# Création depuis un dictionnaire de listes
data_villes = {
    "ville": ["Lille", "Paris", "Lyon", "Marseille"],
    "population": [236000, 2161000, 522000, 870000],
    "superficie_km2": [34.8, 105.4, 47.9, 240.6]
}

df_villes = pd.DataFrame(data_villes)
df_villes
```

```python
# Les commandes réflexes d'inspection :
print("Shape :", df_villes.shape)
print("Types de colonnes :\n", df_villes.dtypes)
df_villes.info()
```
